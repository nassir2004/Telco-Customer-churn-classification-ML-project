# Data understanding

### Telco Customer Churn - Dataset Columns & Business Meaning

| Column | Business Meaning |
| :--- | :--- |
| **customerID** | A unique administrative reference code assigned to each customer account. |
| **gender** | Indicates whether the account holder is male or female. |
| **SeniorCitizen** | Indicates whether the customer belongs to the senior citizen age group (1 for Yes, 0 for No). |
| **Partner** | Refers to whether the customer has a spouse or marital partner. |
| **Dependents** | Specifies whether the customer lives with dependents, such as children or elderly family members. |
| **tenure** | The total number of consecutive months the customer has stayed with the company. |
| **PhoneService** | Shows if the customer is subscribed to a home landline phone service. |
| **MultipleLines** | Specifies whether the customer has additional phone lines connected under their account. |
| **InternetService** | The type of home broadband internet connection provided (DSL, Fiber Optic, or No Internet). |
| **OnlineSecurity** | Indicates if the customer has an add-on service for online threat and malware protection. |
| **OnlineBackup** | Indicates if the customer is subscribed to a cloud storage service for data backup. |
| **DeviceProtection** | Specifies whether the customer has insurance covering hardware failures and damage. |
| **TechSupport** | Shows if the customer subscribes to dedicated, priority technical assistance services. |
| **StreamingTV** | Indicates whether the customer uses the internet connection to stream live television channels. |
| **StreamingMovies** | Shows if the customer utilizes internet bandwidth to stream movies and on-demand video. |
| **Contract** | The duration and structure of the customer agreement (Month-to-month, One-year, or Two-year). |
| **PaperlessBilling** | Indicates if the customer receives digital statements instead of printed paper bills. |
| **PaymentMethod** | The primary payment channel used by the customer (Electronic check, Mailed check, Bank transfer, Credit card). |
| **MonthlyCharges** | The recurring amount billed to the customer each month for all active services. |
| **TotalCharges** | The cumulative amount billed to the customer over the entire duration of their relationship. |
| **Churn** | Indicates whether the customer canceled their subscription and left the company in the past month (Yes/No). |

```python
def clean_total_charges(x):
    if type(x) == float or type(x) == int:
        return x
    elif str(x).strip() == '':
        return 0.0
    else:
        return x

df['TotalCharges'] = df['TotalCharges'].apply(clean_total_charges).astype('float')
df['TotalCharges']
```

### Summary of EDA Findings & Final Key Takeaways
**Goal:** Synthesize core insights from the Exploratory Data Analysis to justify feature selection and guide the machine learning pipeline.

#### **Key Analytical Insights:**

1. **Tenure Impact:** High attrition occurs mainly in the first 12 months (**0-1 Year** segment), after which retention increases significantly.
2. **Contract & Payment Risk:** **Month-to-month** contracts and **Electronic Check** payments are the primary drivers of churn.
3. **Product Engagement:** Customers subscribing to more services (**Total_Services > 3**) show drastically lower churn rates, proving the engineered feature's predictive value.
4. **Service Quality Issues:** **Fiber Optic** users experience a **41.9%** churn rate, signaling potential pricing or service quality issues compared to DSL.
5. **Demographic Risk:** **Senior Citizens** churn at almost double the rate (**41.7%**) of non-seniors (**23.6%**), highlighting a vulnerable customer segment.

# Data Preprocessing for ML

### Check correlation to prevent data leakge

```python
df_encoded = pd.get_dummies(df.drop(columns=['customerID', 'tenure_group'], errors='ignore'), drop_first=True)

churn_corr = df_encoded.corr()['Churn_Yes'].abs().sort_values(ascending=False)

print("Top 10 Features correlated with Churn:")
print(churn_corr.iloc[1:11])
```

### Split Data into Input & Output Features

```python
x = df.drop('Churn', axis= 1)
y = df['Churn']
```

### Check imbalance percentage

```python
y.value_counts(normalize=True)*100
```

### Remapping the target variable to binary format for modeling

```python
y = df['Churn'].map({'No': 0, 'Yes': 1})
y
```

### Numerical Pipeline

- Numerical Pipeline 1

```python
x.select_dtypes(include= 'number').columns
```

```python
scaling_cols = ['tenure', 'MonthlyCharges', 'TotalCharges', 'Total_Services']
scaling_cols
```

```python
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import StandardScaler


rc = StandardScaler()
scaling_pipeline_1 = Pipeline(steps= [('Scaling', rc)])
scaling_pipeline_1
```

### Categorical pipeline

- categorical pipeline 1

```python
ohe_cols = x.select_dtypes(include= 'object').columns.drop('Contract')
ohe_cols
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

ohe_pipeline = Pipeline(steps= [('OHE', ohe)])
ohe_pipeline
```

- categorical pipeline 2

```python
ord_cols = ['Contract']
ord_cols
```

```python
from sklearn.preprocessing import OrdinalEncoder

ord = OrdinalEncoder(categories= [['Month-to-month', 'One year', 'Two year']])

ord_pipeline = Pipeline(steps= [ ('Ordinal Encoder', ord) ])
ord_pipeline
```

```python
x['Contract'].unique()
```

### Assign each column to its corresponding Pipeline

```python
from sklearn.compose import ColumnTransformer

preprocessing = ColumnTransformer(
    transformers=[
        ('Num Pipeline', scaling_pipeline_1, scaling_cols),
        ('OHE Pipeline', ohe_pipeline, ohe_cols),
        ('Ordinal Pipeline', ord_pipeline, ord_cols)
    ],
    remainder='passthrough'  
)

preprocessing
```

### Select Best model

```python
scale_pos_weight = (y == 0).sum() / (y == 1).sum()
scale_pos_weight
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier


models = [
    ('Logistic Regression', LogisticRegression(random_state=42, class_weight='balanced')),
    ('Gaussian NB', GaussianNB()),
    ('SVC', SVC(random_state= 42)),
    ('KNN', KNeighborsClassifier()),
    ('Decision Tree', DecisionTreeClassifier(random_state=42, class_weight='balanced')),
    ('Random Forest', RandomForestClassifier(random_state=42, class_weight='balanced')),
    ('XGBoost', XGBClassifier(random_state=42, scale_pos_weight=scale_pos_weight))
]

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)



for name, model in models:
    model_pipeline = Pipeline(steps=[
        ('Preprocessing', preprocessing),
        ('Model', model)
    ])
    
    cv_result = cross_validate(
        estimator=model_pipeline,
        X=x,
        y=y,
        scoring=['precision', 'recall', 'f1', 'accuracy'],
        cv=skf,
        return_train_score=True,
        n_jobs=-1
    )
    
    print(f"=== {name} ===")
    print(f"Train Recall:    {cv_result['train_recall'].mean() * 100:.2f}%")
    print(f"Test Recall:     {cv_result['test_recall'].mean() * 100:.2f}%")
    print(f"Train Precision: {cv_result['train_precision'].mean() * 100:.2f}%")
    print(f"Test Precision:  {cv_result['test_precision'].mean() * 100:.2f}%")
    print(f"Test F1-Score:   {cv_result['test_f1'].mean() * 100:.2f}%")
    print(f"Test Accuracy:   {cv_result['test_accuracy'].mean() * 100:.2f}%")
    print('-' * 40)
```

### Threshold

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_recall_curve


model_pipeline = Pipeline(steps=[
    ('Preprocessing', preprocessing),
    ('Model', LogisticRegression(random_state=42, class_weight='balanced', max_iter=1000))
])

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

y_proba = cross_val_predict(
    estimator=model_pipeline,
    X=x,
    y=y,
    cv=skf,
    method='predict_proba',
    n_jobs=-1
)[:, 1]
```

```python
y_proba
```

```python
precision, recall, thresholds = precision_recall_curve(y, y_proba)

pr_df = pd.DataFrame({
    'Precision': precision[:-1],
    'Recall': recall[:-1],
    'Threshold': thresholds
})
pr_df
```

```python
pr_df_filtered = pr_df[pr_df['Precision'] >= 0.60]

pr_df_sorted = pr_df_filtered.sort_values(by=['Recall', 'Precision'], ascending=[False, False])

optimum_threshold = pr_df_sorted.iloc[0]['Threshold']

print(f"Optimum Threshold: {optimum_threshold:.4f}")
print(f"Associated Recall:  {pr_df_sorted.iloc[0]['Recall'] * 100:.2f}%")
print(f"Associated Precision: {pr_df_sorted.iloc[0]['Precision'] * 100:.2f}%")
```

##### Not good idea to use this threshold for production, but it can be used for analysis and understanding the trade-offs between precision and recall.

### Hyperparameter tunning by GridSearch

```python
from sklearn.model_selection import GridSearchCV

grid_params = [
    # Logistic Regression Tuning
    {
        'model': [LogisticRegression(random_state=42, class_weight='balanced', max_iter=1000)],
        'params': {
            'Model__C': [0.01, 0.1, 1, 10],
            'Model__solver': ['lbfgs', 'liblinear']
        }
    },
    # Random Forest Tuning
    {
        'model': [RandomForestClassifier(random_state=42, class_weight='balanced')],
        'params': {
            'Model__n_estimators': [100, 200],
            'Model__max_depth': [3, 4, 5, 6],
            'Model__min_samples_leaf': [2, 4, 8]
        }
    },
    # XGBoost Tuning
    {
        'model': [XGBClassifier(random_state=42, scale_pos_weight=scale_pos_weight)],
        'params': {
            'Model__n_estimators': [100, 200],
            'Model__max_depth': [2, 3, 4],
            'Model__learning_rate': [0.01, 0.05, 0.1],
            'Model__subsample': [0.7, 0.8]
        }
    }
]

best_models = {}

for grid in grid_params:
    model_name = grid['model'][0].__class__.__name__
    
    pipeline = Pipeline(steps=[
        ('Preprocessing', preprocessing),
        ('Model', grid['model'][0])
    ])
    
    grid_search = GridSearchCV(
        estimator=pipeline,
        param_grid=grid['params'],
        cv=skf,
        scoring='f1',
        return_train_score=True,
        n_jobs=-1
    )
    
    grid_search.fit(x, y)
    
    best_models[model_name] = grid_search
    
    print(f"=== {model_name} ===")
    print(f"Best Parameters: {grid_search.best_params_}")
    print(f"Best Test F1-Score: {grid_search.best_score_ * 100:.2f}%")
    print('-' * 50)
```

```python
tuned_best_models = [
    ('Random Forest (Tuned)', best_models['RandomForestClassifier'].best_estimator_),
    ('XGBoost (Tuned)', best_models['XGBClassifier'].best_estimator_),
    ('Logistic Regression (Tuned)', best_models['LogisticRegression'].best_estimator_)
]

for name, model_pipe in tuned_best_models:
    cv_final = cross_validate(
        estimator=model_pipe,
        X=x,
        y=y,
        scoring=['precision', 'recall', 'f1', 'accuracy'],
        cv=skf,
        return_train_score=True,
        n_jobs=-1
    )
    
    print(f"=== {name} ===")
    print(f"Train Recall:    {cv_final['train_recall'].mean() * 100:.2f}% | Test Recall:    {cv_final['test_recall'].mean() * 100:.2f}%")
    print(f"Train Precision: {cv_final['train_precision'].mean() * 100:.2f}% | Test Precision: {cv_final['test_precision'].mean() * 100:.2f}%")
    print(f"Test F1-Score:   {cv_final['test_f1'].mean() * 100:.2f}%")
    print(f"Test Accuracy:   {cv_final['test_accuracy'].mean() * 100:.2f}%")
    print('-' * 60)
```

### Select Best model

```python
final_pipeline = best_models['XGBClassifier'].best_estimator_

final_pipeline.fit(x, y)
```
